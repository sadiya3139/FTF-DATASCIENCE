## Assignment 3: Pie, Donut, and Waffle Charts in Python

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
Birds_df=pd.read_csv(r"C:\Users\USER\Downloads\ArewaDS-Machine-Learning-main\ArewaDS-Machine-Learning-main\data\Birds.csv")
```

### Step 3: Prepare the Data

```python
# Count birds by habitat (replace 'Habitat' with your column name if different)
bird_counts = Birds_df["Category"].value_counts()

print(bird_counts)
```


### Pie Chart

```python
plt.figure(figsize=(8, 8))

plt.pie(
    bird_counts,
    labels=bird_counts.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Distribution of Birds by Habitat")

plt.show()
```

### Doughnut chart

```python

plt.figure(figsize=(8, 8))

plt.pie(
    bird_counts,
    labels=bird_counts.index,
    autopct='%1.1f%%',
    startangle=90
)

# Create the donut hole
centre_circle = plt.Circle((0, 0), 0.70, fc='white')
fig = plt.gcf()
fig.gca().add_artist(centre_circle)

plt.title("Donut Chart of Bird Distribution")

plt.tight_layout()
plt.show()
```

### Waffle Chart (Without Extra Libraries)

```python
from pywaffle import Waffle
import matplotlib.pyplot as plt
```

```python
# Total number of squares
total = bird_counts.sum()

# Percentage for each category
percentages = (bird_counts / total * 100).round().astype(int)

# Create a 10 x 10 grid
waffle = np.zeros((10, 10))

current = 0

for i, value in enumerate(percentages):
    for j in range(value):
        row = current // 10
        col = current % 10

        if row < 10:
            waffle[row, col] = i + 1

        current += 1

plt.figure(figsize=(10, 5))

plt.imshow(waffle)

plt.title("Waffle Chart of Bird Distribution")

plt.xticks([])
plt.yticks([])

plt.show()


data = bird_counts.to_dict()

fig = plt.figure(
    FigureClass=Waffle,
    rows=10,
    values=data,
    legend={
        'loc': 'lower left',
        'bbox_to_anchor': (1, 0)
    },
    figsize=(10, 6)
)

plt.title("Bird Distribution Waffle Chart")

plt.show()
```
