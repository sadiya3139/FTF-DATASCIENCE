### Assignment 4 Histogram

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
Birds_df=pd.read_csv(r"C:\Users\USER\Downloads\ArewaDS-Machine-Learning-main\ArewaDS-Machine-Learning-main\data\Birds.csv")
print(Birds_df.head())
```

# Bird Dataset Analysis

Source:
This dataset contains measurements of different bird species including
their wingspan, mass, region, and other biological characteristics.

Objective:
Explore the numerical variables using histograms.

### Histogram 1

Wingspan Distribution

```python
plt.figure(figsize=(8,5))

plt.hist(Birds_df["MaxWingspan"],bins=20)
plt.title("Distribution of Wingspan")
plt.xlabel("MaxWingspan")
plt.ylabel("Frequency")

plt.show()
```

### Histogram 2

Mass Distribution

```python
plt.figure(figsize=(8,5))

plt.hist(Birds_df["MaxBodyMass"], bins=20)

plt.title("Distribution of Mass")
plt.xlabel("BodyMass")
plt.ylabel("Frequency")

plt.show()
```

### Histogram 3
Break length

```python
plt.figure(figsize=(8,5))

plt.hist(Birds_df["MaxLength"], bins=15)

plt.title("Distribution of Beak Length")
plt.xlabel("Beak Length")
plt.ylabel("Frequency")

plt.show()
```

### Histogram 4
Wing length

```python
plt.figure(figsize=(8,5))

plt.hist(Birds_df["MaxWingspan"], bins=15)

plt.title("Distribution of Wing Length")
plt.xlabel("Wing Length")
plt.ylabel("Frequency")

plt.show()
```

### Histogram 5
Tail length

```python
plt.figure(figsize=(8,5))

plt.hist(Birds_df["MaxLength"], bins=15)

plt.title("Distribution of Tail Length")
plt.xlabel("Tail Length")
plt.ylabel("Frequency")

plt.show()
```

# Findings

1. Wingspan appears approximately normally distributed.
2. Most birds have moderate body mass.
3. Beak lengths cluster around the middle values.
4. Wing lengths have few extreme values.
5. Tail lengths vary among bird species.
