```python
plt.title('Max Wingspan in Centimeters')
plt.ylabel('Wingspan (CM)')
plt.tick_params(axis='both',which='both',labelbottom=False,bottom=False)

for i in range(len(Birds_df)):
    x = Birds_df['Name'][i]
    y = Birds_df['MaxWingspan'][i]
    plt.plot(x, y, 'bo')
    if Birds_df['MaxWingspan'][i] > 500:
        plt.text(x, y * (1 - 0.05), Birds_df['Name'][i], fontsize=12)
    
plt.show()
```

### Select Snow Geese

```python
snow_geese = Birds_df[Birds_df["Name"] == "Snow Goose"]

snow_geese.head()
```

### Let's visualize these birds' wingspan by showing a very basic line plot

```python
plt.title('Max Wingspan in Centimeters')
plt.ylabel('Wingspan (CM)')
plt.xlabel('Birds')
wingspan = Birds_df.MaxWingspan 
wingspan.plot()
```

### Line Chart

Average Wingspan by Year

```python
plt.figure(figsize=(10,5))

plt.plot(
    snow_geese["Year"],
    snow_geese["Wingspan"],
    marker='o'
)

plt.title("Snow Goose Wingspan Over the Years")
plt.xlabel("Year")
plt.ylabel("Wingspan (cm)")
plt.grid(True)

plt.show()
```

### Scatter Plot

Body Mass vs Wingspan

```python
plt.figure(figsize=(8,6))

plt.scatter(
    snow_geese["Mass"],
    snow_geese["Wingspan"]
)

plt.title("Mass vs Wingspan of Snow Geese")
plt.xlabel("Mass (g)")
plt.ylabel("Wingspan (cm)")

plt.show()
```

### Bar Chart

```python
Number of Snow Geese by Region

region_counts = snow_geese["Region"].value_counts()

plt.figure(figsize=(8,5))

region_counts.plot(kind="bar")

plt.title("Snow Geese by Region")
plt.xlabel("Region")
plt.ylabel("Number of Birds")

plt.show()
```

### Group By Conservation Status(Bar Chart)

```python
grouped=Birds_df.groupby("ConservationStatus")[["MinLength","MaxLength","MinBodyMass","MaxBodyMass"]].mean()
grouped.plot(kind="bar",figsize=(10,6))
plt.title("Birds Measurement by Conservation status")
plt.xlabel("conservation status")
plt.ylabel("Average Value")
plt.legend(title="Metrics")
plt.show
```

### Group by Conservation Status(compare only lengths)

```python
grouped=Birds_df.groupby("ConservationStatus")[["MinLength","MaxLength"]].mean()
grouped.plot(kind="bar",figsize=(10,6) )
plt.title("Birds Measurement by Conservation status  comparing only Length")
plt.xlabel("conservation status")
plt.ylabel("Average Value")
plt.legend(title="Metrics")
plt.show
```

### Group by Conservation Status(Compare onlyBody Mass)

```python
grouped=Birds_df.groupby("ConservationStatus")[["MinBodyMass","MaxBodyMass"]].mean()
grouped.plot(kind="bar",figsize=(10,6) )
plt.title("Birds Measurement by Conservation status  comparing only Body Mass")
plt.xlabel("conservation status")
plt.ylabel("Average Value")
plt.legend(title="Metrics")
plt.show
```

### Group by Category(Bar chart)

```python
grouped=Birds_df.groupby("Category").size()
grouped.plot(kind="bar",figsize=(10,6),color="green")
plt.title("Birds Measurement by Category")
plt.xlabel("Category")
plt.ylabel("count")
plt.xticks(rotation=45)
plt.show
```
