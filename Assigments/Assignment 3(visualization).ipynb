```python
import pandas as pd
import matplotlib.pyplot as plt
print(pd.__version__)
Birds_df=pd.read_csv(r"C:\Users\USER\Downloads\ArewaDS-Machine-Learning-main\ArewaDS-Machine-Learning-main\data\Birds.csv")
print(Birds_df.head())
```

```python
grouped=Birds_df.groupby("Name")[["Name","ScientificName"]].size()
grouped.plot(kind="bar",figsize=(10,6))
plt.title("Birds Names by Scientific Names")
plt.xlabel("ScientificName")
plt.ylabel("Names")
plt.legend(title="Metrics")
plt.show
```

### 2a. Group By Conservation Status(Bar Chart)

```python
grouped=Birds_df.groupby("ConservationStatus")[["MinLength","MaxLength","MinBodyMass","MaxBodyMass"]].mean()
grouped.plot(kind="bar",figsize=(10,6))
plt.title("Birds Measurement by Conservation status")
plt.xlabel("conservation status")
plt.ylabel("Average Value")
plt.legend(title="Metrics")
plt.show
```

### 2b. Group by Conservation Status(compare only lengths)

```python
grouped=Birds_df.groupby("ConservationStatus")[["MinLength","MaxLength"]].mean()
grouped.plot(kind="bar",figsize=(10,6) )
plt.title("Birds Measurement by Conservation status  comparing only Length")
plt.xlabel("conservation status")
plt.ylabel("Average Value")
plt.legend(title="Metrics")
plt.show
```

### 2c.Group by Conservation Status(Compare onlyBody Mass)

```python
grouped=Birds_df.groupby("ConservationStatus")[["MinBodyMass","MaxBodyMass"]].mean()
grouped.plot(kind="bar",figsize=(10,6) )
plt.title("Birds Measurement by Conservation status  comparing only Body Mass")
plt.xlabel("conservation status")
plt.ylabel("Average Value")
plt.legend(title="Metrics")
plt.show
```

### 2d.Birds Distribution by Conservation Statu(Pie Chart)

```python
grouped=Birds_df.groupby("ConservationStatus").size()
grouped.plot(kind="pie",autopct="%1.1f%%")
plt.title("Birds Distribution by Conservation Status")
plt.ylabel("")
plt.show
```

### 3a.Group by Category(Bar chart)

```python
grouped=Birds_df.groupby("Category").size()
grouped.plot(kind="bar",figsize=(10,6),color="green")
plt.title("Birds Measurement by Category")
plt.xlabel("Category")
plt.ylabel("count")
plt.xticks(rotation=45)
plt.show
```

### 3b.Group By Category:order (Bar chart)

```python
plt.title('Max Wingspan in Centimeters')
plt.ylabel('Wingspan (CM)')
plt.tick_params(axis='both',which='both',labelbottom=False,bottom=False)

for i in range(len(Birds_df)):
    x = Birds_df['Name'][i]
    y = Birds_df['MaxWingspan'][i]
    plt.plot(x, y, 'bo')
    if Birds_df['MaxWingspan'][i] > 500:
        plt.text(x, y * (1 - 0.05), Birds_df['Name'][i], fontsize=12)
    
plt.show()
```
