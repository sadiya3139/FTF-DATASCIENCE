Case Study 1
Unauthorized Sale of Nigerian Citizens’ Personal Data for ₦100: The XpressVerify & AnyVerify Incident (2024)
Data Ethics Case Study  •  Arewa Data Science ML Program
1. DATA ETHICS CHALLENGE IDENTIFIED
Privacy & Unauthorized Data Use / Lack of Informed Consent
This case study focuses on two interconnected data ethics violations: unauthorized access to personal data and commercialization of data without consent. These challenges map to the Deon Checklist categories of Privacy and Consent — which require that individuals should know and agree to how their data is collected, stored, and used, especially when it involves sensitive identity information.

2. REAL-WORLD INCIDENT
Background
Nigeria’s National Identity Management Commission (NIMC) collects biometric and personal data from citizens and issues the National Identification Number (NIN). The Central Bank of Nigeria manages the Bank Verification Number (BVN) system. These two identifiers are among the most sensitive pieces of personal data a Nigerian citizen possesses — linked to bank accounts, passport, driving licence, voter card, and tax records.

What Happened
In March 2024, investigative journalism outlet FIJ (Foundation for Investigative Journalism) published an exposé revealing that a private website, XpressVerify.com.ng, had unrestricted access to the National Identity Database and had monetized the recovery of NINs and personal information — with vendors across Nigerian cities using it to look up any citizen’s full name, address, phone number, and photograph.
After XpressVerify was taken down, Paradigm Initiative discovered another actor: AnyVerify.com.ng, which had been operating since November 2023. AnyVerify offered NIN, BVN, virtual NIN, driving licence, international passport, company details, TIN, PVC, and phone numbers — all sold to any interested party for as little as ₦100 per data request.
The scale of access was alarming: AnyVerify was visited 567,990 times in February 2024 and 188,360 times in April 2024 alone.

Data Ethics Questions This Raises
•	Consent: Did any Nigerian citizen agree to have their NIN, BVN, address, or photograph sold to strangers for ₦100?
•	Purpose Limitation: Data was collected for identity management and governance — not commercial resale.
•	Data Security: Who authorized third-party API access to the national identity database?
•	Accountability: Who is responsible — the websites, NIMC, or the NDPC for failing to enforce the law?
•	Transparency: Citizens were never informed their data was accessible to private commercial actors.

Harms and Unintended Consequences
•	Identity theft, financial fraud, and physical safety risks — including targeting by burglars, kidnappers, or terrorists using home address data.
•	Erosion of public trust in Nigeria’s banking and identity systems.
•	Breach of driver’s licence and passport data creating criminal exploitation pathways.
•	The cheap marketplace for identity credentials dramatically lowered barriers for SIM-swap fraud and account impersonation.

Potential Solutions
•	Revoke and audit all third-party API access to NIMC and CBN databases.
•	Enforce the Nigeria Data Protection Act (NDPA) 2023 with meaningful penalties.
•	Require Data Privacy Impact Assessments (DPIAs) before any third-party API integration.
•	Establish a data subject notification system so citizens are alerted when their data is queried.
•	Create a licensed and publicly verifiable whitelist of data processors with strict usage limits.
Paradigm Initiative filed a landmark public-interest case at the Federal High Court in Abuja in July 2024, seeking compensation for affected data subjects, investigation of all data leaks, and publication of findings.

Laws Violated	NDPA 2023; Section 37, Constitution of Nigeria

```python
Case Study 2
“Pay or Be Shamed”: Nigeria’s Digital Loan App Crisis & the Weaponization of Personal Data (2021–2024)
Data Ethics Case Study  •  Arewa Data Science ML Program
1. DATA ETHICS CHALLENGE IDENTIFIED
Unauthorized Secondary Use of Data / Consent Violation / Data as a Weapon
This case study focuses on the deliberate misuse of personal data collected for one stated purpose (loan disbursement) to inflict psychological harm on borrowers as a debt recovery tactic. It also touches on data minimization (collecting far more data than necessary) and human dignity — themes central to the Deon Checklist under Privacy, Consent, and Unintended Use.

2. REAL-WORLD INCIDENT
Background
Nigeria’s traditional banking system is largely inaccessible to its population. Only about 2% of Nigerian adults can access bank loans. Into this gap entered hundreds of digital lending apps — promising instant loans via smartphone. By clicking “Accept” on a lengthy terms document, borrowers unknowingly handed over access to their entire phone: contacts, photos, location, messages, and call logs.

What Happened
When borrowers were even slightly late on repayment, apps deployed their harvested data as a weapon. Digital Money Lenders (DMLs) used forgery, impersonation, and privacy breaches to torment defaulters — sending harassing voice notes and defamatory messages to borrowers’ family members, friends, employers, and community members.
One victim, Roseline, described how a minor salary delay triggered a cascade of abuse: “Every night, I cried, I lost weight, I couldn’t eat, and nothing was appealing to me.” She was pushed to the brink of suicide.
Another victim, Jumoke, watched her debt balloon to ₦926,385 — over 51 times her original loan amount. Apps sent defamatory messages calling her a fraudster, and printed and posted flyers with her picture to publicly shame her. The harassment drove her to attempt to take her own life.
Between 2021 and 2023, the Federal Competition and Consumer Protection Commission (FCCPC) received over 11,000 complaints about harassment, data abuse, and unethical recovery tactics by digital lenders. Some apps — like Fly Pay — even impersonated the EFCC and Nigeria Immigration Service, sending fake arrest warrants and travel ban notices to defaulters.

Data Ethics Questions This Raises
•	Consent: Did borrowers meaningfully agree to their contacts being used as leverage? Clicking “Accept” on an opaque 40-page ToS does not constitute informed consent.
•	Purpose Limitation: Data collected for “loan disbursement and identity verification” was deliberately repurposed to inflict shame and psychological harm.
•	Data Minimization: Why does a ₦6,000 loan require access to a borrower’s full contact list, location data, and photographs?
•	Human Dignity: Broadcasting defamatory messages to a borrower’s family and employer is not an accident — it is a designed feature of the business model.

Harms and Unintended Consequences
•	Mental Health: Nigerians lost jobs, sank into depression, and in some cases died by suicide due to unrelenting harassment.
•	Medical/Psychological: Health experts documented sleep disorders, appetite loss, shame, depression, high blood pressure, and anxiety disorders.
•	Social Harm: Defamatory messages broadcast to employers, pastors, and parents cause irreversible reputational damage.
•	Systemic Harm: The harassment triggered counter-movements including Facebook groups where victims plotted ways to defraud loan apps — a second layer of criminal activity.
•	Predatory Economics: Effective annual interest rates on some loans reached 3,600%; borrowers sometimes received only ₦59,000 out of an approved ₦100,000.

Potential Solutions
•	The NDPC investigated over 400 privacy breach cases and drafted the NDPA-GAID, mandating privacy-by-design and requiring platforms to deny access to those who use them for violations.
•	The FCCPC released a list of 173 approved digital lenders required to disclose interest rates and data access practices.
•	In April 2023, Google updated Play Store policy to prohibit loan apps from accessing contacts, photos, videos, precise location, and call logs.
•	Mandatory Data Privacy Impact Assessments before any loan app is approved for distribution.
•	Criminalization of contact-list harvesting as a debt recovery method under the Cybercrimes Act.

3. SUMMARY TABLE
Challenge	Unauthorized secondary use of data; consent violation; data weaponization
Data Misused	Phone contacts, photos, location, call logs
Who Was Harmed	Millions of Nigerian borrowers; their families and contacts
Root Cause	Predatory business model deliberately designed around data exploitation
Laws Violated	NDPA 2023, NDPR 2019, Cybercrimes Act, FCCPC Guidelines
Scale	11,000+ FCCPC complaints (2021–2023); 400+ active NDPC investigations (2024)
Human Cost	Depression, job loss, social ruin, and documented cases of suicide attempts

4. RELATED RESOURCES
Daily Trust – Digital Loan Sharks Investigation (November 2023):  https://dailytrust.com/digital-loan-sharks-shaming-recovery-techniques-pushing-nigerians-to-the-brink/
Melting Pot Africa – Jumoke’s Story (October 2024):  https://www.meltingpot.africa/nigeria-loan-app/
PUNCH – NDPC Investigating 400+ Privacy Breach Cases (March 2024):  https://punchng.com/commission-probes-400-cases-of-privacy-breach-in-online-loan-apps/
Rest of World – Nigerians Fight Back in Facebook Groups (2023):  https://restofworld.org/2023/facebook-groups-debt-shaming-nigeria/
```
