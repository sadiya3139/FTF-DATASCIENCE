# Day 10 - Loops

# For loop

```python

for i in range(5):
    print(i)
```

# Loop over list

```python
fruits = ["apple", "banana", "mango"]
for fruit in fruits:
    print(fruit)
```

# While loop

```python
i = 0
while i < 5:
    print(i)
    i += 1
```

# Break and continue

```python
for i in range(5):
    if i == 3:
        break
    print(i)

for i in range(5):
    if i == 3:
        continue
    print(i)
```
