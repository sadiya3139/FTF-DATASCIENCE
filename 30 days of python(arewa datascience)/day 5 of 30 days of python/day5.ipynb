```python
fruits[0] = "apple"
print(fruits)
```

# Add items

```python
fruits.append("grape")
print(fruits)
```

# Insert

```python
fruits.insert(2, "pineapple")
print(fruits)
```

# Remove

```python
fruits.remove("orange")
print(fruits)
```

# Copy

```python
new_list = fruits.copy()
print(new_list)
```

```python
fruits.reverse()
print(fruits)
```
