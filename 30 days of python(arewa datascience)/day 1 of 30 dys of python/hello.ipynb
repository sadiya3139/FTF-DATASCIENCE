# 4. Check data type

```python
print(type(10))
print(type(9.8))
print(type(3.14))
print(type(4 - 4j))
print(type(['Asabeneh', 'Python', 'Finland']))
print(type("Sadiya"))
print(type("Rabiu"))
print(type("Nigeria"))
#OR
# Check Data Types
def check_data_types():
    data_list = [
        10,
        9.8,
        3.14,
        4 - 4j,
        ['Asabeneh', 'Python', 'Finland'],
        "Sadiya",
        "Rabiu",
        "Nigeria"
    ]
    
    for data in data_list:
        print(data, "->", type(data))
```
