# Day 8 - Dictionaries

# Creating dictionary

```python
person = {
    'first_name': 'Sadiya',
    'last_name': 'Rabiu',
    'age': 25,
    'country': 'Nigeria',
    'skills': ['Python', 'AI', 'Cybersecurity']
}

print(person)
print(person['first_name'])
print(person.get('age'))
```

# Adding items

```python
person['city'] = 'Abuja'
print(person)
```

# Updating

```python
person['age'] = 26
print(person)
```

# Removing

```python
person.pop('city')
print(person)
del person['last_name']
print(person)
```


# Keys, values, items

```python
print(person.keys())
print(person.values())
print(person.items())
```
