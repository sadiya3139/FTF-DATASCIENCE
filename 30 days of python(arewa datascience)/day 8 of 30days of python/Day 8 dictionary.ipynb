# Day 8 - Dictionaries

# Creating dictionary

```python
person = {
    'first_name': 'Sadiya',
    'last_name': 'Rabiu',
    'age': 25,
    'country': 'Nigeria',
    'skills': ['Python', 'AI', 'Cybersecurity']
}

print(person)
print(person['first_name'])
print(person.get('age'))
```

# Adding items

```python
person['city'] = 'Abuja'
print(person)
```

# Updating

```python
person['age'] = 26
print(person)
```

# Removing

```python
person.pop('city')
print(person)
del person['last_name']
print(person)
```


# Keys, values, items

```python
print(person.keys())
print(person.values())
print(person.items())
```

# another example

# Dog dictionary

```python
dog = {}
dog['name'] = "Bingo"
dog['color'] = ['Brown']
dog['breed'] = "Local"
dog['legs'] = 4
dog['age'] = 3
print(len(dog))

print(dog['legs'], type(dog['legs']))

dog['color'].append('black')

print(list(dog.keys()))
print(list(dog.values()))

print(dog.items())

dog.pop('age')
```

# Student dictionary

```python
student = {
    'first_name': 'Sadiya',
    'last_name': 'Rabiu',
    'gender': 'Female',
    'age': 28,
    'marital_status': 'Single',
    'skills': ['Python', 'AI'],
    'country': 'Nigeria',
    'city': 'kano',
    'address': 'Hotoro'
}

print(len(student))

print(student['skills'], type(student['skills']))

student['skills'].append('Cybersecurity')

print(list(student.keys()))
print(list(student.values()))

print(student.items())

student.pop('age')
```
