# Day 7 -Sets

# Creating sets

```python
st1 = set()
st2 = {1, 2, 3, 4}
st3 = {1, 2, 2, 3, 4}
print(st2)
print(st3)  # duplicates removed
```

# Adding items

```python
st2.add(5)
print(st2)
```

# Removing items

```python
st2.remove(3)
print(st2)
st2.discard(10)  # no error if not present
```

# Set operations

```python
a = {1, 2, 3}
b = {3, 4, 5}

print(a.union(b))
print(a.intersection(b))
print(a.difference(b))
print(a.symmetric_difference(b))
```
