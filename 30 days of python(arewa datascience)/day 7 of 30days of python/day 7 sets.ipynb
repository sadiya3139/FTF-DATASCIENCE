# Day 7 -Sets

# Creating sets

```python
st1 = set()
st2 = {1, 2, 3, 4}
st3 = {1, 2, 2, 3, 4}
print(st2)
print(st3)  # duplicates removed
```

# Adding items

```python
st2.add(5)
print(st2)
```

# Removing items

```python
st2.remove(3)
print(st2)
st2.discard(10)  # no error if not present
```

# Set operations

```python
a = {1, 2, 3}
b = {3, 4, 5}

print(a.union(b))
print(a.intersection(b))
print(a.difference(b))
print(a.symmetric_difference(b))
```

# Sets Exercises

```python
it_companies = {'Facebook', 'Google', 'Microsoft', 'Apple', 'IBM', 'Oracle', 'Amazon'}
A = {19, 22, 24, 20, 25, 26}
B = {19, 22, 20, 25, 26, 24, 28, 27}
age = [22, 19, 24, 25, 26, 24, 25, 24]
```

# Level 1

```python
print(len(it_companies))

it_companies.add('Twitter')

it_companies.update(['Tesla', 'Netflix'])

it_companies.remove('IBM')
# Difference remove vs discard
# remove → error if not found
# discard → no error
```


# Level 2

```python

print(A.union(B))
print(A.intersection(B))
print(A.issubset(B))
print(A.isdisjoint(B))

print(A.union(B))
print(B.union(A))

print(A.symmetric_difference(B))

del A
del B
```

# Level 3

```python
age_set = set(age)
print(len(age), len(age_set))
```
