# DAY 12 - LIST COMPREHENSION

```python
numbers = [-4, -3, -2, -1, 0, 2, 4, 6]
```

# Negative & zero

```python

print([i for i in numbers if i <= 0])
```

# Flatten list

```python
list_of_lists = [[[1,2,3]], [[4,5,6]], [[7,8,9]]]
flattened = [num for sub in list_of_lists for row in sub for num in row]
print(flattened)
```

# Tuple pattern

```python
pattern = [(i, 1, i, i**2, i**3, i**4, i**5) for i in range(11)]
print(pattern)
```

# Countries transformation

```python
countries = [[('Finland', 'Helsinki')],
             [('Sweden', 'Stockholm')],
             [('Norway', 'Oslo')]]

result = [[c[0].upper(), c[0][:3].upper(), c[1].upper()] for sub in countries for c in sub]
print(result)

```

# Convert to dictionary

```python
dict_result = [{'country': c[0].upper(), 'city': c[1].upper()} for sub in countries for c in sub]
print(dict_result)
```

# Names concatenation

```python
names = [[('Asabeneh', 'Yetayeh')],
         [('David', 'Smith')],
         [('Donald', 'Trump')],
         [('Bill', 'Gates')]]

print([f"{f} {l}" for sub in names for (f, l) in sub])
```
