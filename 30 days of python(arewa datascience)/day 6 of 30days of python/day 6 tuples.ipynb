# creating tuple

```python
tpl1 = ()
tpl2 = (1, 2, 3)
tpl3 = ('Sadiya', 'Rabiu', 'Nigeria')
tpl4 = (1, 2, 3, ('a', 'b'))

print(tpl2)
print(tpl3)
print(tpl4)
```

# Accessing elements

```python
print(tpl3[0])
print(tpl4[3][1])
```

# Slicing

```python
print(tpl2[1:])
print(tpl3[:2])
```

# Unpacking


```python
name, surname, country = tpl3
print(name, surname, country)
```

# Tuple operations

```python
print(len(tpl2))
print(tpl2 + (4, 5))
print(tpl2 * 2)
```

# 2. Food tuples

```python
fruits = ("apple", "banana", "mango")
vegetables = ("carrot", "tomato")
animal_products = ("milk", "meat")

food_stuff_tp = fruits + vegetables + animal_products
```

# 3. Convert to list

```python
food_stuff_lt = list(food_stuff_tp)
```

# 4. Middle item

```python
mid = len(food_stuff_lt) // 2
print("Middle item:", food_stuff_lt[mid])
```

# 5. First & last 3

```python
print("First 3:", food_stuff_lt[:3])
print("Last 3:", food_stuff_lt[-3:])
```
