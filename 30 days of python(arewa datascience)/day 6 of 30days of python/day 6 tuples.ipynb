# DAY 6 -Tuple

# Creating tuple

```python
tpl1 = ()
tpl2 = (1, 2, 3)
tpl3 = ('Sadiya', 'Rabiu', 'Nigeria')
tpl4 = (1, 2, 3, ('a', 'b'))

print(tpl2)
print(tpl3)
print(tpl4)
```

# Accessing elements

```python
print(tpl3[0])
print(tpl4[3][1])
```

# Slicing

```python
print(tpl2[1:])
print(tpl3[:2])
```

# Unpacking


```python
name, surname, country = tpl3
print(name, surname, country)
```

# Tuple operations

```python
print(len(tpl2))
print(tpl2 + (4, 5))
print(tpl2 * 2)
```
